#if-elif : when we have multiple conditions to check but it is wan't INDEPENDANT

```python
# creating calculator program using if-elif
num1=float(input("Enter the num1 :"))
num2=float(input("Enter the num2 :"))
operator=input("Enter the operator :")
if operator=="+" :
  result=num1+num2
  print("Addition is :",result)
elif operator=="-" :
  result=num1-num2
  print("Substraction is :",result)
elif operator=="*" :
  result=num1*num2
  print("Multiplication is :",result)
elif operator=="/" :
  result=num1/num2
  print("Dividation is :",result)
else :
  print("Invalid Operator")
```

```python
# if age>18 and do ypu have voting id card : yes==>eligible
age=int(input("Enter your age :"))
if age>=18 :
  voting_id=input("Do you have voting id card(yes/no) :").lower()
  if voting_id=="yes":
    print("Eligible")
  else :
    print("plz bring voting id card")
else :
  print("Not Eligible")
```

```python
#password checker
#conditions
#len<=8
#first char==>uppercase:
#last char==digit
password=str(input("enter the password:"))
if len(password)<=8 :
  if password[0].isupper()==True:
    if password[-1].isdigit()==True:
      print("Correct Password")
    else:
      print("last char must be digit")
  else :
    print("First char must be in uppercase")
else :
  print("wrong password!")





```

```python
# Q: Online shopping app
# condition
# purchase amount
# if purchase amount>1000 and membership=yes, discount=10% , membership="no" ,discount=8%
# if purchase amount between 500 t0 1000 then and membership=yes, then discount=5%,membership=no then discount=3%
# purchase amount<500 and membership=yes the discount=2%
# purchase amount<500 and membership=No the discount=0%
purchase_amount=float(input("Enter Your Purchase Amount :"))
Membership=input("Enter your Membership (yes/No):")
if purchase_amount>=1000:
  if Membership=="yes":
    discount=purchase_amount*0.1
  else:
    discount=purchase_amount*0.08
elif 1000>purchase_amount>=500:
  if Membership=="yes":
    discount=purchase_amount*0.05
  else:
    discount=purchase_amount*0.03
elif purchase_amount<500:
  if Membership=="yes":
    discount=purchase_amount*0.02
  else:
    discount=0
final_amount=purchase_amount-discount
print("This is final amount :",final_amount)





```
