#### Student score analysis

```python
# store student names and their marks in a dictionary and then calculate avg marks for each students.
d1={"Affan":[86,56,75,89],
    "Tanmay":[53,52,99,63],
    "Vijay":[58,89,57,75]}
for name,score in d1.items():
  avg=sum(score)/len(score)
  print(f"{name}={avg}")
```

#### Nested Dictionary

```python
sales={
    "jan":{"prod A":500,"prod B":600},
    "feb":{"prod A":400,"prod B":700},
    "mar":{"prod A":1000,"prod B":800}
}
# sum of sales for each month
for key,value in sales.items():
    print(f"{key} : {sum(value.values())}")
```

```python
# Q2. Extract prime numbers from list
p=[1,3,2,5,7,13,2,6,8]
# prime number 1>
# prime
prime=[]
for i in p:
  if i>1:
    for divider in range(2,i):
      if i % divider==0:
        break
    else:
      prime.append(i)
prime

```

```python
range1=int(input("Enter your range"))
prime=[]
for i in range(2,range1+1): # 0 and 1 are not prime number
  if i>1:
    for divider in range(2,i):
      if i % divider==0:
        break
    else:
      prime.append(i)
prime
```

```python
# Q3 : extract words where len>5 and first character in uppercase
l=["data","python","science","Data","Data Analysis"]
result = []

for word in l:
    if len(word) > 5 and word[0].isupper():
        result.append(word)
print("Filtered words:", result)
```
