```python
# create list of squares from numbers 1 to 100
squares=[]
for i in range(1,101):
  squares.append(i**2)
print(squares)
```

```python
# Using List Comphrension
# syntax :
  # variable name =[output for loop contional]
squares=[i**2 for i in range(1,101)]
print(squares)
```

```python
s= " i am learning python"
# upper each words in a string and print output in the list format
l=[]
for i in s.split():
  l.append(i.upper())
l
```
