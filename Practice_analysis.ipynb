```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
eda=pd.read_csv("/content/sales_dataset.csv")
eda
```

```python
eda.head()
```

```python
eda.dtypes
```

```python
# how to convert date dtype
eda['Date']=pd.to_datetime(eda['Date'])
```

```python
eda.dtypes
```

```python
eda.isnull
```

```python
#Handle missing data: fill or drop where appropriate (e.g., mean for numerical, mode for
#categorical)
eda.isnull().sum()
```

```python
eda["Unit_Price"] = eda["Unit_Price"].fillna(eda["Unit_Price"].mean())
eda
```

```python
eda.isnull().sum()
```

```python
eda["Discount"] = eda["Discount"].fillna(eda["Discount"].mean())
eda
```

```python
eda.isnull().sum()
```

```python
eda["Shipping_Cost"] = eda["Shipping_Cost"].fillna(eda["Shipping_Cost"].mean())
eda
```

```python
# mode for Customer_Name
eda["Customer_Name"] = eda["Customer_Name"].fillna(eda["Customer_Name"].mode()[0])
eda
```

```python
eda.isnull().sum()
```

```python
eda
```

```python
#Identify and correct invalid entries (e.g., negative quantities or sales)
```

```python
#Create new derived columns:
#Net_Sales = Total_Sales - (Total_Sales * Discount)
eda["Net_Sales"] = eda["Total_Sales"] - (eda["Total_Sales"] * eda["Discount"])
eda
```

```python
eda["Profit_Estimate"] = eda["Net_Sales"] - eda["Shipping_Cost"]
eda
```

```python
eda.duplicated
```

```python
eda.drop_duplicates
```

```python
#Generate summary statistics for numerical and categorical variables.
eda.describe()
```

```python
#Region with the highest total sales
eda.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False)
```

```python
#Product category with the lowest average profit
eda.groupby("Product_Category")["Profit_Estimate"].mean().sort_values(ascending=True)
```

```python
#Month with maximum total sales
eda["Month"] = eda["Date"].dt.month
eda
```

```python
#Sales by region (bar chart)
import seaborn as sna
import matplotlib.pyplot as plt
sna.barplot(x="Region", y="Total_Sales", data=eda)
```

```python
#• Average profit by product category
sna.barplot(x="Product_Category", y="Profit_Estimate", data=eda)
```

```python
#Save the cleaned dataset as 'cleaned_sales_data.csv'.
eda.to_csv("cleaned_sales_data.csv", index=False)
```
