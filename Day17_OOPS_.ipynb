#Encapsulation : we can do in the 3 ways
- 1. public members(variables) : define as a ("self.name")
- 2. protected members : define as a ("self._name")
- 3. private members : define as a ("self.__name")

```python
class person:
  def __init__(self,name,age):
    self.name=name
    self._age=age
    self.__city="pune"
  def show_info(self):
    print("Name :",self.name)
    print("Age :",self._age)
    print("City :",self.__city)
```

```python
class subclass(person):
  def inherit(self):
    print("Name :",self.name)
    print("Age :",self._age)
    print("City :",self.__city)
```

```python
object_class=subclass("Affan",22)
object_class.inherit()
```

```python
# protected
class base:
  def __init__(self):
    self._a=20
class derived(base):
  def protected(self):
    print(f"Calling protected member of base class :{self._a}")
```

```python
ob=derived()
ob.protected()
```

```python
# private
class base:
  def __init__(self):
    self.__a=20
  def show(self):
    print(self.__a)
class derived(base):
  def protected(self):
    print(f"Calling protected member of base class :{self.__a}")
  def f2(self):
    print(self.__a)
```

```python
o=derived()
o.protected()
o.f2()
```

###Method Overriding : we can use same method name in parent class and its sub class to do different task

```python
class bird:
  def intro(self):
    print("There are different types of birds")
  def flight(self):
    print("most of the birds fly and some cannot")
class parrot(bird):
  def flight(self):
    print("parrot can fly")
class pengium(bird):
  def flight(self):
    print("Pengium don't fly")
```

```python
b=bird()
b.intro()
b.flight()
p=parrot()
p.flight()
pe=pengium()
pe.flight()
```

# Inheritance : Types of Inheritance
- 1.single Inheritance
- 2.multiple Inheritance

- 5.Hybrid Inheritance


####1.single Inheritance

```python
class A: #parent class
  def feature1(self):
    print("Feature 1 working")
  def feature2(self):
    print("Feature 2 working")
class B(A): #child class
  def feature3(self):
    print("Feature 3 working")
  def feature4(self):
    print("Feature 4 working")

```

```python
single=B()
single.feature1()
single.feature4()
```

####2.multiple Inheritance

```python
class A: #parent class 1
  def feature1(self):
    print("Feature 1 working")
  def feature2(self):
    print("Feature 2 working")
class B: #parent class 2
  def feature3(self):
    print("Feature 3 working")
  def feature4(self):
    print("Feature 4 working")
class c(A,B): # child class
  def feature5(self):
    print("Feature 5 working")
  def feature6(self):
    print("Feature 6 working")
```

```python
multiple=c()
multiple.feature1()
multiple.feature3()
multiple.feature5()
```

####3.multi-level Inheritance

```python
class A: #parent class
  def feature1(self):
    print("Feature 1 working")
  def feature2(self):
    print("Feature 2 working")
class B(A): # Intermidate class
  def feature3(self):
    print("Feature 3 working")
  def feature4(self):
    print("Feature 4 working")
class c(B): # child class
  def feature5(self):
    print("Feature 5 working")
  def feature6(self):
    print("Feature 6 working")
```

```python
multilevel=c()
multilevel.feature1()
multilevel.feature3()
multilevel.feature5()
```

####4.Hierarchical Inheritance

```python
class A: # Main parent class
  def feature1(self):
    print("Feature 1 working")
  def feature2(self):
    print("Feature 2 working")
class B(A): # child class 1
  def feature3(self):
    print("Feature 3 working")
  def feature4(self):
    print("Feature 4 working")
class c(A): # child class 2
  def feature5(self):
    print("Feature 5 working")
  def feature6(self):
    print("Feature 6 working")
```

```python
Hierarchical=c()
Hierarchical.feature1()
Hierarchical.feature5()
b2=B()
b2.feature1()
b2.feature3()
```

####5.Data Abstraction
- hide implementation part of software from the end user
- we can't create object for Abstract class
- How to achieve data abstraction
  - step 1 : import abc package and from that package import ABC class and abstractmethod(fn)
  - step 2 : inherit this ABC class in the 2nd class
  - steo 3 : create 1 and 2 abstract method in that class using abstractclassmethod
- Abstract method : A method/function ==> only declare but not implement
- Abstract class : A class which contains 1 or more abstract method
- concrete class : A class which doesn't contain any abstract method

```python
# concrete class
class my_class:
  def study(self):
    print("studying python")
```

```python
o=my_class()
o.study()
```

```python
# Abstract class
from abc import ABC , abstractclassmethod
class my_class1(ABC):
  def study(self):
    print("i am studying python")
  @abstractclassmethod
  def mock(self):
    pass
```

```python
# we can't create object for Abstract class
obj=my_class1()
obj.study()
```

```python
import math
class shape(ABC):
  def main(self):
    print("This is main function")
  @abstractclassmethod
  def area(self):
    pass
  @abstractclassmethod
  def perimeter(self):
    pass
class circle(shape):
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    print("The area of circle is :",math.pi*self.radius*self.radius)
  def perimeter(self):
    return 2*math.pi*self.radius
class rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    print("The area of rectangle is :",self.length*self.width)
  def perimeter(self):
    return 2*(self.length+self.width)
```

```python
c=circle(5)
c.area()
c.main()
```

```python
c.perimeter()
```

```python
r=rectangle(2,3)
r.area()
r.main()
```

```python
r.perimeter()
```

```python
# Task : create car class with 3 abstract method engine,breaks,wheels and two normal methods is turn_on_lights,play_music
# and then create a subclass tesla that provides implementation for all abstract methods
from abc import ABC , abstractclassmethod
class car(ABC):
  @abstractclassmethod
  def engine(self):
    pass
  @abstractclassmethod
  def breaks(self):
    pass
  @abstractclassmethod
  def wheels(self):
    pass
  def is_turn_on_lights(self):
    print("lights are on")
  def is_play_music(self):
    print("music is playing")
class tesla(car):
  def engine(self):
    print(" electric engine ")
  def breaks(self):
    print(" Auto breaks")
  def wheels(self):
    print("Auto Rotating wheels")

```

```python
t=tesla()
t.engine()
t.breaks()
t.wheels()
```

```python
t.is_turn_on_lights()
t.is_play_music()
```

####Super Method() : we can directly call methods from the abstractclass to its sub-class using super method()

```python
import math
class shape(ABC):
  def main(self):
    print("This is main function")
  @abstractclassmethod
  def area(self):
    pass
  @abstractclassmethod
  def perimeter(self):
    pass
class rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    print("The area of rectangle is :",self.length*self.width)
    super().main()
  def perimeter(self):
    return 2*(self.length+self.width)
    super().main()
```

```python
# area is use with print function that reason the super() method work.
r=rectangle(2,3)
r.area()
```

```python
# In the perimeter we use ruturn fun that reason super() does't work
r.perimeter()
```
