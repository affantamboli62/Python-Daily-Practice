```python
# print numbers from 1 to 100 using while loop
num=int(input("Enter your number :"))
while num<=100:
  print(num)
  num=num+1
```

```python
# take current bill amount from user
# run while loop until bill!=0
# calculate total bill amount
bill=float(input("Enter your First bill :"))
total=bill
while bill!=0:
  bill=float(input("Enter your Next bill :"))
  total=total+bill
total
```

```python
# using for loop
bill_list=[]
t=0
for i in range(1,3):
  bill=float(input("Enter your bill amount :"))
  bill_list.append(bill)
  t=t+bill
print(bill_list)
print(t)
```

# Task : Calculate Total votes for each person

```python
# vote="A" for manu
# vote="M" for alex
# mcount=0
# acount=0
# vote= "plz enter your vote"
# election = "Quit"

vote=""  #initilization
mcount=0
acount=0
while vote!="quit":
  vote=str(input("Plz Enter your vote(M/A) :"))
  if vote=="M":
    mcount+=1
  elif vote=="A":
    acount+=1
  else:
    print("invalid code")
print("Manu vote count :",mcount)
print("Alex vote count :",acount)


```

# break statement

```python
# Take distance and time value from user
# and then calculate speed=d/t
# if distance entered is negative then print msg : "invalid distance,plz enter positive distance"
# and then break while loop

# initilization
# choice="yes"
# stop while loop when choice="no"

choice="yes"
while choice!="no":
  distance=float(input("Enter your distance :"))
  if distance<0:
    print("invalid distance , plz enter positive distance")
    break
  else:
    time=float(input("Enter your time :"))
    speed=distance/time
  print(speed)
  choice=input("Enter your choice :")


```

# Continue statement

```python
choice="yes"
while choice!="no":
  distance=float(input("Enter your distance :"))
  if distance<0:
    print("invalid distance , plz enter positive distance")
    break
  else:
    time=float(input("Enter your time :"))
    speed=distance/time
  print(speed)
  choice=input("Enter your choice :")
```
