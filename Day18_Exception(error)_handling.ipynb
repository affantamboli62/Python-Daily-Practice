#python try with else clause :      
- in some situations you might want to run a certain block of code if the code inside try run without any errors. for these cases, you can use optional else keyword with the try statement.
- else clause runs only if **no exceptions **occurs inside try clause
- will go in else , if no error or exception in try
- otherwise if exception(error) in try then will go in except and will skip else clause

```python
try:
  a=1
  b=0
  e=a/b
  p=a*b
  y=a+b

except:
  print("divide by non zero otherwise uh will get infite number")

print(a)
print(b)

print("this is last line")
```

```python
class login_error(Exception):
  pass
try:
  email="affan62"
  password=1234
  enter_email=str(input("Enter a email :"))
  enter_password=int(input("Enter a password :"))
  if email==enter_email and password==enter_password:
    print("Login Successfully")
  else:
    raise login_error("Wrong Email")
except login_error as ex:
  print(ex)

else:
  print("This will run when try is correct")

finally:
  print("You are trying to login with speedup")
```

```python
#Task : using try , except , else , finally to validate students marks
#- 1.Accept marks input from the user
#- 2.Raise a error if marks are not between 0 to 100
#- 3.Handle invalid input(non-numeric)
#- 4.Display the grades based on marks
#- 5.Show completion msg: marks entry competed
class invalid_marks(Exception):
  pass
try:
  marks=int(input("Enter a marks :"))
  if marks<0 or marks>100:
    raise invalid_marks("your marks is not correct")

except invalid_marks as invalid_marks:
  print(invalid_marks)

else:
  print("your marks is :",marks)

finally:
  print("marks entry completed")
```
