new HTML file

```python
HTML2=open("/content/Example.html")
```

```python
cleaned=bs(HTML2,"html")
cleaned
```

```python
sales_record=cleaned.find_all("div",class_="sales-record")
sales_record
```

```python
order_ids=[]
date=[]
for data in sales_record:
  s=data.find("span",class_="order-id")
  s2=data.find("span",class_="date")
  order_ids.append(s.get_text())
  date.append(s2.get_text())
print(order_ids)
print(date)
```

```python
customers=[]
citys=[]
for data in sales_record:
  s=data.find("span",class_="customer")
  s2=data.find("span",class_="city")
  customers.append(s.get_text())
  citys.append(s2.get_text())
print(customers)
print(citys)
```

```python
products=[]
quantities=[]
for data in sales_record:
  s=data.find("span",class_="product")
  s2=data.find("span",class_="quantity")
  products.append(s.get_text())
  quantities.append(s2.get_text())
print(products)
print(quantities)
```

```python
units=[]
totals=[]
for data in sales_record:
  s=data.find("span",class_="unit-price")
  s2=data.find("span",class_="total")
  units.append(s.get_text())
  totals.append(s2.get_text())
print(units)
print(totals)
```
