#Bar chart==> vertical column chart (x="Text",y="Numeric")

```python
bigmart.plot(kind="bar",x="Item_Fat_Content",y="Item_Outlet_Sales")
```

```python
grp=bigmart.groupby("Item_Fat_Content")["Item_Outlet_Sales"].sum()
grp
```

```python
grp.plot(kind="bar",x="Item_Fat_Content",y="Item_Outlet_Sales")
```

```python
import seaborn as sns
sns.barplot(x="Item_Fat_Content",y="Item_Outlet_Sales",data=bigmart)
```

#Scatterplot==> bivariate analysis
- numeric to numeric
- To check correlation between 2 variables

```python
df=pd.DataFrame({"emp_id":[101,201,103,203,105],
                      "exp":[1,2,3,4,5],
                      "salary":[10000,20000,30000,40000,50000],
                      "age":[-500,23,54,19,500]

})
df
```

```python
df.plot(kind="scatter",x="exp",y="salary")
```

```python
df.corr(numeric_only=True)
```

```python
df["age"].plot(kind="box")
```

```python
df.plot.pie(y="salary")
```
