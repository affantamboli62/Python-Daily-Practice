```python
d={"affan":22,"tanmay":45,"vijay":33}
sorted(d)   # sorted key wise
```

```python
d.items()
```

```python
# keywise sorting
sorted_dict=dict(sorted(d.items(),key=lambda x:x[0]))
sorted_dict
```

```python
# valuewise sorting
sorted_dict=dict(sorted(d.items(),key=lambda x:x[1]))
sorted_dict
```

```python
# reverse
sorted_dict=dict(sorted(d.items(),key=lambda x:x[1],reverse=True))
sorted_dict
```

```python
words=["apple","banana","kiwi","cherry"]
sorted_list=[]
sorted_list=sorted(words,key=lambda x:len(x))
sorted_list
```

```python
# using def fun
def sort_len(list):
  return sorted(list,key=lambda x:len(x))
sort_len(["apple","banana","kiwi","cherry"])


```
