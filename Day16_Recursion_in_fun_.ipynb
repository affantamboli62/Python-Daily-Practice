#Recursion : Repeat
- function call itself

```python
# factorial
# without function
n=int(input("enter a num :"))
fact=1
for i in range(1,n+1):
  fact=fact*i
print(fact)
```

```python
# factorial
# with function
def factorial1(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  return fact
factorial1(5)
```

with Recursion

```python
def factorial1(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial1(n-1)
factorial1(5)
```

#Fibonacci Series: Addition of previous term

```python
#Without function
#[0,1,1,2,3,5,8,13.....]
fibo=[0,1]
#first 10 numbers
for i in range(1,9):
  next_term=fibo[-1]+fibo[-2]
  fibo.append(next_term)
print(fibo)
```

```python
#With function
def fib(n):
  fibo=[0,1]
#first 10 numbers
  for i in range(1,n-1):
    next_term=fibo[-1]+fibo[-2]
    fibo.append(next_term)
  return fibo
fib(10)
```

```python
# with recursion
# with recursion
def fib_recursive(n):
  if n <= 1:
    return n
  else:
    return(fib_recursive(n-1) + fib_recursive(n-2))
fib_recursive(10)
```

```python
# with recursion
def fib_recursive(n):
  if n <= 1:
    return n
  else:
    return(fib_recursive(n-1) + fib_recursive(n-2))

# Print the first 10 Fibonacci numbers using the recursive function
n_terms = 10
if n_terms <= 0:
   print("Plese enter a positive integer")
else:
   print("Fibonacci sequence:")
   for i in range(n_terms):
       print(fib_recursive(i))
```

```python
# Task : sum of natural number using recursion
# with Recursion
def sum_natural_numbers(n):
  if n <= 1:
    return n
  else:
    return n + sum_natural_numbers(n-1)
sum_natural_numbers(10)
```

```python
# Task 2: reverse string using recursion
def reverse_string(s):
  if len(s) == 0:
    return s
  else:
    return s[-1]+reverse_string(s[:-1])
reverse_string("data")

# How the recursion work :
# "a"+reverse_string("dat")
# "t"+reverse_string("da")
# "a"+reverse_string("d")
# "d"+reverse_string()
```
