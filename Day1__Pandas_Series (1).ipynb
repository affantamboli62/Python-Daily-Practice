#Series Arithmatic Operations:

###Difference Between List and Series:
- List - Hetroginious ==> Multiple Data Type
  - l1=[1,2.3,"Affan"]
  - l2=["Tanmay",20,30]
  - For Arithmetic Operations Between Two List we need Lopp's
- Series - Homoginious ==> Single Data Type
  - s=[1,2,3.5,"Affan",7.8.9]==> ["1","2","3.5","Affan","7.8","9"]
  - prefernces
    - str>float>int
  - Arithmetic operators ==> faster ==> due to same data types
  - There is no need of loops for Arithmetic opeartions
  - larger data

```python
l1=[45,1,2.3]
l2=[20,30,45]
l1+l2

# indexwise Addition

print(l1[0]+l2[0])
print(l1[1]+l2[1])
print(l1[2]+l2[2])
```

```python
# Using For loop
l1=[45,1,2.3]
l2=[20,30,45]
l1+l2
addition=[]
for i in range(len(l1)):
  addition.append(l1[i]+l2[i])
print(addition)
```
