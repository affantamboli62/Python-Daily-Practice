```python
data=pd.DataFrame({
    "city":["Mumbai","Nashik","Pune","Mumbai","Pune"],
    "temp":[10,5,12,7,4]
})
data
```

```python
data.groupby("city")["temp"].max()
```

```python
# add column max temp in orignal dataframe
data["Max_Temp"]=data.groupby("city")["temp"].transform("max")
data
```

```python
# Avg temp
data["Avg_Temp"]=data.groupby("city")["temp"].transform("mean")
data
```

##apply()

```python
# divide temp by 10
data["Temp/10"]=data["temp"].apply(lambda x:x/10)
data
```

```python
def fn(n):
  return n/10
```

```python
data["2Temp/10"]=data["temp"].apply(fn)
data
```

```python
# count no of characters in each city
data["city_len"]=data["city"].apply(len)
data
```

```python
# if max_temp > 10 , "yes","no"
data["hot_temp"]=data["Max_Temp"].apply(lambda x:"yes" if x>10 else "no")
data
```
