```python
# 4.Write a program that handles multiple exceptions — ValueError and
# ZeroDivisionError — when dividing two user inputs.
try:
  num1=int(input("Enter a number one :"))
  num2=int(input("Enter a number two :"))
  c=num1/num2
except ValueError:
  print("Invalid input")
except ZeroDivisionError:
  print("Cannot divide by zero")
else:
  print("Division Sucessful :",c)
```

```python
#15.Create a DataFrame of 5 students with marks in two subjects.Find the mean marks for each subject using .mean().
#Find the maximum and minimum marks using .max() and .min().

import pandas as pd
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Math': [85, 90, 78, 92, 88],
    'Science': [70, 85, 91, 75, 80]
}
df= pd.DataFrame(data)
df

mean_marks = df[['Math', 'Science']].mean()
print(mean_marks)

max_marks = df[['Math', 'Science']].max()
print(max_marks)

min_marks = df[['Math', 'Science']].min()
print(min_marks)
```
