##fillna==parameter==> method="bfill" or "ffill" ==> backward fill and forward fill ==> used in time series data

- 12-1-2025     - 500
- 13-1-2025     - 1000
- 14-1-2025     - nan
- 15-1-2025     - 2000
- 16-1-2025     - 2500
