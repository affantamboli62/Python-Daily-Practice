```python
# sqrt
def squareroot(a):
  s=a**0.5
  print(s)
squareroot(64)
```

```python
# abs
def absolute():
  num=int(input("Enter a number :"))
  num_str=str(num)
  if num_str[0]=="-":
    print(int(num_str[1:]))
  else:
    print(num)
absolute()
```

```python
# count
def countee(data,targetvalue):
  c=0
  for item in data:
    if item==targetvalue:
      c=c+1
  print(c)
countee("affan","a")


```

```python
# len
def length(iterable):
  l=0
  for i in iterable:
    l=l+1
  return l
length("affan")


```

###Difference : print & return
- print() :display output

- return :
    - display(),reuse fun inside another fun
    - in one function we can write only one return
    - a fun stops executing as soon as it reaches a return statement


```python
def square(a):
  s=a**2
  print(s)
square(4)
```

```python
def cube(a):
  print(square(a)*a)
cube(12)
```

```python
def square_return(a):
  s=a**2
  return s
square_return(4)
```

```python
def cube(a):
  print(square_return(a)*a)
cube(12)
```

return : a fun stops executing as soon as it reaches a return statement

```python
def square_return(a):
  s=a**2
  print("i am affan")
  return s
square_return(4)
```

```python
def square_return(a):
  s=a**2
  return s
  print("i am affan")
square_return(4)
```

```python
def square_return(a):
  s=a**2
  return s
  return "i am affan"
square_return(4)
```

```python
def square_return(a):
  s=a**2
  return "i am affan"
  return s
square_return(4)
```
